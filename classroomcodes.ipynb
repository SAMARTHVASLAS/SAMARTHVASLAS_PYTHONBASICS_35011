```python
# Each expression is evaluated directly
print('10 + 5 =', 10 + 5)
print('10 - 5 =', 10 - 5)
print('10 * 5 =', 10 * 5)
print('10 / 5 =', 10 / 5)
print('11 // 5 =', 11 // 5)
print('11 % 5 =', 11 % 5)
print('2 ** 3 =', 2 ** 3)
```

```python
print('left associative division:', 6 / 3 / 2)
print('explicit grouping:', (6 / 3) / 2)
print('right associative power:', 2 ** 3 ** 2)
print('same grouping:', 2 ** (3 ** 2))
print('different grouping:', (2 ** 3) ** 2)
```

```python

# Direct string comparisons
print("'a' > 'A':", 'a' > 'A')
print("'A' < 'a':", 'A' < 'a')
print("'abc' > 'ABC':", 'abc' > 'ABC')
print("'cat' < 'cater':", 'cat' < 'cater')
print("'same' == 'same':", 'same' == 'same')

# ord() shows the numeric code point of one character
print('A:', ord('A'))
print('a:', ord('a'))
print('0:', ord('0'))
print('!:', ord('!'))
```

```python
# Truth-table cases written explicitly
print('not True  =', not True)
print('not False =', not False)
print('True and True   =', True and True)
print('True and False  =', True and False)
print('False and True  =', False and True)
print('False and False =', False and False)
print('True or True    =', True or True)
print('True or False   =', True or False)
print('False or True   =', False or True)
print('False or False  =', False or False)
```

```python

# Short-circuiting prevents evaluation of 10 / 0 here.
# If it were evaluated, division by zero would raise an error.
print(False and (10 / 0))
print(True or (10 / 0))

# Ordinary Boolean expressions
age = 21
has_id = True
print((age >= 18) and has_id)
print((age < 18) or (not has_id))
```

```python
# Shorthand assignment practice
value = 20
value += 5
value *= 2
value //= 3
value %= 7
print('final value:', value)
```

```python

# Straight-line solution using only Chapter 1 concepts
product = 'Notebook'
price = 120.0
quantity = 3
discount_rate = 0.10
subtotal = price * quantity
discount = subtotal * discount_rate
total = subtotal - discount
print('Product:', product)
print('Subtotal:', round(subtotal, 2))
print('Discount:', round(discount, 2))
print('Total:', round(total, 2))
print('Notebook' < 'Pencil')
```
